```python
# matplotlib 基础
import matplotlib.pyplot as plt


x = range(-10000, 10000)
y = [i**3 for i in x]

# 　设置坐标轴显示中文
plt.rcParams['font.sans-serif'] = ['Noto Sans CJK JP']

# 设置画布
plt.figure(figsize=(10, 6))
plt.plot(x, y)  # 绘制折线图----没有画布时，默认创建一个画布

# 设置详细信息
plt.title('X^3 图像')
plt.xlabel('X 轴')
plt.ylabel('Y 轴')
plt.show()
```

- 画布: plt.figure(figsize=(10, 6))
- 标题: plt.title('标题')
- 图例: plt.legend(['标签1', '标签2'])
- 网格: plt.grid(True)
- 轴标签: plt.xlabel('x轴标签')
- 轴标签: plt.ylabel('y轴标签')
- 轴刻度: `plt.xticks([0, 1, 2, 3, 4])`
- 轴刻度: `plt.yticks([0, 1, 2, 3, 4])`
  -  内部参数[]是x,y轴的数据的索引  
- 轴刻度标签: plt.xtickslabels(['标签1', '标签2', '标签3', '标签4', '标签5'])
- 轴刻度标签: plt.ytickslabels(['标签1', '标签2', '标签3', '标签4', '标签5'])
- 轴范围: plt.xlim(0, 4)
- 轴范围: plt.ylim(0, 4)

# 创建子图
- 在单个画布上显示多个图表
- 函数：`plt.subplot(nrows, ncols, figsize(int, int), dpi(int))`---->元组(子图对象, 轴对象)
  - `nrows`：行数----(画布上子图的行数)
  - `ncols`：列数----(画布上子图的列数)
  - `figsize`：画布大小
  - `dpi`：图表分辨率


```python
from calendar import c
from pydoc import text

from matplotlib import axes
from matplotlib.pylab import f
countries = ['中国', '美国', '日本', '韩国', '英国', '法国', '意大利', '印度', '德国']
oil_storage = [400,500,125,50,250,54,65,84,120]

figure,axes = plt.subplots(1, 2, figsize=(20, 10),dpi = 100)

axes_oil = axes[0]
axes_oil.set_title('石油储量',fontsize=20)
axes_oil.bar(countries, oil_storage)
axes_oil.set_xlabel('国家',fontsize=20)
axes_oil.set_ylabel('储量',fontsize=20)
# axes_oil.set_xticklabels(countries, rotation=45) # rotation=45 旋转45度, 避免标签重叠
axes_oil.grid(True)

# 世界人口
# '中国', '美国', '日本', '韩国', '英国', '法国', '意大利', '印度', '德国'
# 对应人口数量为：
population = [1400,320,125,50,250,54,65,1409,120]
axes_population = axes[1]
axes_population.pie(population, labels=countries, autopct='%1.1f%%',textprops={'fontsize':20})
axes_population.set_label(countries)
axes_population.set_title('世界人口',fontsize=20)
axes_population.legend(loc='lower center',ncol=5,bbox_to_anchor=(0.5, -0.15),fontsize=20)
```
