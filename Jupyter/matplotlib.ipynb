```python
# matplotlib 基础
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm

x = range(-10000, 10000)
y = [i**3 for i in x]

#　设置坐标轴显示中文
plt.rcParams['font.sans-serif'] = ['Noto Sans CJK JP']

# 设置画布
plt.figure(figsize=(10, 6))
plt.plot(x, y) # 绘制折线图----没有画布时，默认创建一个画布

# 设置详细信息
plt.title('X^3 图像')  
plt.xlabel('X 轴')
plt.ylabel('Y 轴')
plt.show()
```
