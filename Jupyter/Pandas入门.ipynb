```python
# Pandas数据的读取和写入
# read_XXXX     读取xxxx数据
# to_XXXX       写入xxxx数据
# import pandas as pd
df = pd.read_csv('Data/opgg_tier.csv')
df
# df['总场数'] = df['胜场'] + df['负场']

# # 调整列顺序
insert_tag = df.columns.tolist().index('胜场')
df.insert(insert_tag, '总场数', df['胜场'] + df['负场'])
# df = df[['#', '召唤师', 'LP', '胜场', '负场', '总场数', '胜率']]
# # 调整列名
col_names_list = df.columns.tolist()
col_names_list[df.columns.get_loc('#')] = '排名'
df.columns = col_names_list
df.to_csv('Data/opgg_tier_adjust.csv', index=False)  # 参数：（文件名路径，index是否包含索引）
df
```

```python
# 数据清洗
# 缺失值,重复值,异常值,数据一致性

# 1. 缺失值
df.isnull()
# 1.1 缺失值统计
df.isnull().sum()
# 1.2 删除缺失值
df.dropna()         # 行删除    修改后的数据作为返回值
df.dropna(axis=1)   # 列删除 默认值为0，表示行删除
# 1.3 填充缺失值
df.fillna(0)        # 填充0
df.ffill()          # 前向填充_用上一行的值填充缺失值
df.bfill()          # 后向填充_用下一行的值填充缺失值


# 2. 重复值
df.duplicated()     # 默认判断所有列数据是否重复
df.duplicated(subset=['LP'])  # 只判断LP列数据是否重复  # subset 用于指定要判断的列
# 2.1 删除重复值
# 可选参数keep:'first'(默认:保留第一个重复值),False(删除所有重复值),'last'(保留最后一个重复值)
df.drop_duplicates(subset=['召唤师'], keep=False)
```

```python
# 3. 异常值-格式处理
# 3.1 查看异常值



df[df['召唤师'].str.contains('#')]
df[df['LP'] < 0]
# 3.2 删除异常值
df.drop(df[df['LP'] < 0].index)  # 可选参数 inplace=True 表示在原数据上进行修改
# # 3.3 修复异常值
temp = df[df['召唤师'].str.contains('#')]['召唤师'].str.split(
    '#', expand=True)  # expand=True 表示切分后返回一个DataFrame
if temp.empty == False:
    print('temp非空')
    df['召唤师'] = temp[0]
    insert_tag = df.columns.tolist().index('召唤师') + 1
    df.insert(insert_tag, '后缀', temp[1])
else:
    print('temp为空')

df.to_csv('Data/opgg_tier_adjustID.csv', index=False)
# 定义一个空的DataFrame
hero_temp = pd.DataFrame()
if hasattr(df['常用英雄'], 'str'):
    hero_temp = df['常用英雄'].str.split('/', expand=True)
if hero_temp.empty == False:
    print('hero_temp非空')
    df_hero = df.drop(columns=['常用英雄'])
    df_hero[[0, 1, 2]] = [None, None, None]
    insert_tag = df_hero.columns.tolist().index('段位')
    new_columns = df_hero.columns.tolist()
    new_columns = new_columns[:insert_tag] +new_columns[-3:]+ new_columns[insert_tag:-3:]
    df_hero = df_hero[new_columns]
    for i in range(len(new_columns)):
        if new_columns[i] in range(3):
            new_columns[i] = ('常用英雄', new_columns[i]+1)
        else:
            new_columns[i] = (new_columns[i], '')
    df_hero.columns = pd.MultiIndex.from_tuples(new_columns)
  
#     # df[[('常用英雄', '英雄2'),('常用英雄', '英雄3')]] = temp
#     # df.drop(columns=['常用英雄'], inplace=True)
    df_hero[[('常用英雄',1),('常用英雄',2),('常用英雄',3)]] = hero_temp
else:
    print('hero_temp为空')
# hero_temp.iloc[45:55:]


df_hero.fillna('无', inplace=True)
df_hero.to_csv('Data/opgg_tier_adjustHero.csv', index=False)
df_hero.loc[45:55:]
```

```python
# 数据排序与分组
# 排序语法:
# df.sort_values(by='列名', ascending=Fasle)
# 可选参数 ascending：False 表示降序排序,默认值为True
# by = 可以不写，直接写列名即可
df.sort_values('总场数', ascending=False)
# 多列排序(排序优先级，左高右低)
df.sort_values(by=['胜场', '负场'], ascending=[False, True])

# 分组语法:
# df.groupby('列名:分组依据')['列名：分组统计'].size()
# 统计方法：
# size()：统计每个分组的样本数量
# mean()：计算每个分组的均值
# median()：计算每个分组的中位数
# std()：计算每个分组的标准差
# var()：计算每个分组的方差
# max()：计算每个分组的最大值
# min()：计算每个分组的最小值
# sum()：计算每个分组的总和
# count()：计算每个分组的非缺失值数量
# agg(方法1,方法2,方法3...)：对每个分组应用多个统计方法
df.groupby('胜场',sort=False)[['排名', 'LP', '胜场', '负场', '总场数']].agg(['mean','sum'])

```
